# Conceito Canny:
 Algoritmo robusto que combina múltiplas etapas para detectar bordas verdadeiras, seguindo três critérios:
 1. Baixa taxa de erro
 2. Boa localização
 3. Resposta única por borda

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carregar imagem e converter para tons de cinza
img = cv2.imread('edificios.jpg')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# Pré-processamento: Suavização com filtro Gaussiano
img_suavizada = cv2.GaussianBlur(img_gray, (17, 17), 0)
```

```python
# Aplicar detector de bordas Canny
canny_result = cv2.Canny(img_suavizada, 40, 60)
```

```python
# Visualização comparativa
plt.figure(figsize=(15, 5))

plt.subplot(121)
plt.imshow(img_suavizada, cmap='gray')
plt.title('Imagem Suavizada')
plt.axis('off')

plt.subplot(122)
plt.imshow(canny_result, cmap='gray')
plt.title('Bordas Detectadas (Canny)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Salvar resultado
cv2.imwrite('canny_result.jpg', canny_result)
print("Resultado Canny salvo como 'canny_result.jpg'")
```
