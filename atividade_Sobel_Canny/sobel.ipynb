```python
## Conceito Sobel:
# Técnica clássica da família de filtros derivativos que realça bordas estimando variações luminosas
# significativas nos pixels.
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carregar imagem e converter para tons de cinza
img = cv2.imread('edificios.jpg')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# Aplicar operador de Sobel
sobelX = cv2.Sobel(img_gray, cv2.CV_64F, 1, 0)
sobelY = cv2.Sobel(img_gray, cv2.CV_64F, 0, 1)

sobelX = np.uint8(np.absolute(sobelX))
sobelY = np.uint8(np.absolute(sobelY))

sobel_combined = cv2.bitwise_or(sobelX, sobelY)
```

```python
# Visualização dos resultados
plt.figure(figsize=(15, 10))

plt.subplot(131)
plt.imshow(sobelX, cmap='gray')
plt.title('Sobel Direção X')
plt.axis('off')

plt.subplot(132)
plt.imshow(sobelY, cmap='gray')
plt.title('Sobel Direção Y')
plt.axis('off')

plt.subplot(133)
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Combinado')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Salvar resultado
cv2.imwrite('sobel_result.jpg', sobel_combined)
print("Resultado Sobel salvo como 'sobel_result.jpg'")
```
